To fix issues, I created a for if else loop where if a line starts with an ">", the loop splits the line and gets the first item or what is at position zero, which is our gene id, and writes it into an outfile. Then, if the if statemet doesn't apply, the else takes the sequence line and and counts the g and c and count and writes it into an outfile. This creates a file with two columns, one with the gene sequence and one with the gc content. Below is that code... 

````python
with open("IntroBiolComp-2026/Python/DataFiles/Turkey_transcripts_15.fasta", "r") as infile, open ("turkey_gc_content.txt", "w") as outfile:
    for line in infile: 
        line = line.rstrip()
        if line.startswith(">"):
            gene_id = line.split()[0]
            outfile.write(gene_id+"\t")
        else:
            seq=line
            gc = (seq.count("G") + seq.count("C")) / len(seq)
            outfile.write(str(gc)+"\n")
````
