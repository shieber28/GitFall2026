Here you will find my code for the fizz buzz game for numbers 1 to 100

```python
for i in range (1, 101):
    if i % 3 == 0:
        print ("fizz")
    if i % 5 == 0:
        print ("buzz")
    if i % 3 == 0 and i % 5 == 0:
        print ("fissbuzz") 
```

While writing the code I wanted to make sure that if none of the statements applied that nothing printed out, so I googled and learned that you can use pass under an else statement so that nothing prints out, but then I also learned after that an else statement isn't even necessary and the code runs the same even without it. I ran the code both ways and it created the same output!
